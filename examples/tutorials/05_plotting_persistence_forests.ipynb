# Plot a Persistence Forest

Build a forest from a 2D point cloud, then draw it with `plot_persistence_forest`.
This star-shaped example produces several visible branches. Run the cells in order.

```python
from pathlib import Path
import sys

# Allow running from the repository root or examples/tutorials.
repo = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "persforest" / "PersistenceForest.py").exists()), None)
if repo is not None and str(repo) not in sys.path:
    sys.path.insert(0, str(repo))

import numpy as np
import matplotlib.pyplot as plt
from persforest import PersistenceForest

rng = np.random.default_rng(7)
theta = np.linspace(0, 2 * np.pi, 650, endpoint=False)
radius = 1 + 0.55 * np.cos(5 * theta) + rng.normal(0, 0.008, theta.size)
points = radius[:, None] * np.column_stack((np.cos(theta), np.sin(theta)))
forest = PersistenceForest(points, print_info=False)
```

## From points to a forest

Height represents filtration value; horizontal position just arranges the branches.
`min_tree_span` hides tiny trees without changing the underlying forest.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)
axes[0].scatter(*points.T, s=4, color="#334155")
axes[0].set(aspect="equal", title="Star-shaped point cloud")
forest.plot_persistence_forest(
    ax=axes[1], show=False, min_tree_span=0.01, title="Persistence forest",
)
plt.show()
```

## Match branches to a vertical barcode

Use `coloring="bars"` in both panels to identify the same persistence intervals by color.
`orientation="vertical"` puts the barcode on the same filtration axis as the forest.
Here `min_bar_length` hides short intervals, and `nodes="critical"` marks roots, junctions, and leaves.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5), sharey=True, constrained_layout=True)
min_length = 0.005
forest.plot_persistence_forest(
    ax=axes[0], show=False, min_bar_length=min_length,
    coloring="bars", nodes="critical", title="Forest",
)
forest.plot_barcode(
    ax=axes[1], orientation="vertical", min_bar_length=min_length,
    coloring="bars", sort="length", title="Barcode", ylabel="",
    tight_layout=False,
)
plt.show()
```

A few useful options:

- `min_branch_span` removes short side offshoots while keeping each tree's trunk.
- `nodes="none"` hides markers; `nodes="all"` shows every visible node.
- `branch_angle=90` makes curved branches leave the trunk orthogonally.

Thresholds use filtration units—squared distances for the default alpha complex. All filters default to zero.
